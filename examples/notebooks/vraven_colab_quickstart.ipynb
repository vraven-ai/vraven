# VRAVEN quickstart

Run VRAVEN in Google Colab without cloning the repository or uploading a model. This notebook installs the public PyPI package, checks the environment and creates one deterministic HTML explanation report.

```python
%pip install -q vraven
```

```python
import subprocess
import sys

subprocess.run([sys.executable, "-m", "vraven.cli", "doctor"], check=True)
```

```python
from pathlib import Path

report_path = Path("/content/vraven-demo.html")
subprocess.run(
    [
        sys.executable,
        "-m",
        "vraven.cli",
        "demo",
        "--mode",
        "gradient",
        "--report-format",
        "html",
        "--output",
        str(report_path),
    ],
    check=True,
)
print(f"Created {report_path} ({report_path.stat().st_size / 1_000_000:.1f} MB)")
```

## Download the report

The report is self-contained. Download it and open it in a browser to inspect the explanation, metadata and stated limitations.

```python
from google.colab import files

files.download(str(report_path))
```
